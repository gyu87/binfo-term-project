```python
import os
from pathlib import Path

# 1. 분석 디렉토리 설정
data_dir = Path("/Users/gyudong/clipseq_project/data")
data_dir.mkdir(parents=True, exist_ok=True)

# 2. SRA ID 리스트
sra_ids = ["SRR11838741", "SRR11838742"]

# 3. 다운로드 및 변환
for sra_id in sra_ids:
    print(f"▶ Downloading: {sra_id}")
    os.system(f"prefetch {sra_id} --output-directory {data_dir}")

    sra_file = data_dir / sra_id / f"{sra_id}.sra"
    
    print(f"▶ Converting to FASTQ: {sra_id}")
    os.system(f"fasterq-dump {sra_file} --split-files -O {data_dir} --threads 4")
    
    print(f"▶ Compressing FASTQ: {sra_id}")
    os.system(f"gzip -f {data_dir}/{sra_id}_*.fastq")

    print(f"▶ Cleaning up .sra file: {sra_id}")
    os.remove(sra_file)
```

```python
# 필요한 패키지 불러오기
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import pearsonr

# 파일 경로 (현재 노트북과 같은 경로일 경우 ./ 사용)
npz_path = './replicate_coverage_10kbp.npz'

# npz 파일 불러오기
data = np.load(npz_path)
read_counts = data['data']  # shape: (n_bins, 2)

# log2(Read Count + 1) 변환
log_counts = np.log2(read_counts + 1)
x = log_counts[:, 0]
y = log_counts[:, 1]

# 산점도 그리기
plt.figure(figsize=(6, 6))
plt.scatter(x, y, s=1, color='red', alpha=0.6)

# x = y 대각선 추가
min_val = min(x.min(), y.min())
max_val = max(x.max(), y.max())
plt.plot([min_val, max_val], [min_val, max_val], 'k--', linewidth=1)

# 축 라벨 및 제목
plt.xlabel("Rep1 log2(Read Count+1)", fontsize=12)
plt.ylabel("Rep2 log2(Read Count+1)", fontsize=12)
plt.title("Pearson Correlation", fontsize=14)

# Pearson 상관계수 계산 및 텍스트 표시
r, _ = pearsonr(x, y)
plt.text(min_val + 0.5, max_val - 1, f"PCC = {r:.3f}", fontsize=12)

# 레이아웃 조정 및 저장
plt.tight_layout()
plt.savefig("replicate_correlation_custom.png", dpi=300)
plt.savefig("replicate_correlation_custom.pdf")

# 노트북 안에서 바로 이미지 확인
plt.show()

```
