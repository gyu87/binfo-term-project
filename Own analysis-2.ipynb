```python
import os
from pathlib import Path

# 1. 분석 디렉토리 설정
data_dir = Path("/Users/gyudong/clipseq_project/data")
data_dir.mkdir(parents=True, exist_ok=True)

# 2. SRA ID 리스트
sra_ids = ["SRR11838741", "SRR11838742"]

# 3. 다운로드 및 변환
for sra_id in sra_ids:
    print(f"▶ Downloading: {sra_id}")
    os.system(f"prefetch {sra_id} --output-directory {data_dir}")

    sra_file = data_dir / sra_id / f"{sra_id}.sra"
    
    print(f"▶ Converting to FASTQ: {sra_id}")
    os.system(f"fasterq-dump {sra_file} --split-files -O {data_dir} --threads 4")
    
    print(f"▶ Compressing FASTQ: {sra_id}")
    os.system(f"gzip -f {data_dir}/{sra_id}_*.fastq")

    print(f"▶ Cleaning up .sra file: {sra_id}")
    os.remove(sra_file)
```
