````
def run_entropy_pipeline(pileup_file, output_bedgraph):
    import pandas as pd
    import numpy as np
    import re
    from collections import Counter

    # 1. load
    pileup = pd.read_csv(pileup_file, sep='\t', names=['chrom', 'pos', '_ref', 'count', 'basereads', 'quals'])

    # 2. clean
    toremove = re.compile('[<>$*#^]')
    pileup['matches'] = pileup['basereads'].apply(lambda x: toremove.sub('', x))

    # 3. count bases
    def count_bases(match_str):
        base_counts = Counter(match_str.upper())
        return [base_counts.get(base, 0) for base in 'ACGT']
    base_counts = pileup['matches'].apply(count_bases)
    pileup[['A', 'C', 'G', 'T']] = pd.DataFrame(base_counts.tolist(), index=pileup.index)

    # 4. entropy
    def shannon_entropy(row):
        counts = np.array([row['A'], row['C'], row['G'], row['T']])
        total = counts.sum()
        if total == 0:
            return 0.0
        probs = counts / total
        return -np.sum([p * np.log2(p) for p in probs if p > 0])
    pileup['entropy'] = pileup.apply(shannon_entropy, axis=1)

    # 5. save bedGraph
    pileup['start'] = pileup['pos'] - 1
    pileup['end'] = pileup['pos']
    bedgraph = pileup[['chrom', 'start', 'end', 'entropy']]
    bedgraph.to_csv(output_bedgraph, sep='\t', header=False, index=False)
````
